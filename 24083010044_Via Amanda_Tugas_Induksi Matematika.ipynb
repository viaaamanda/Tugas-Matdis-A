```python
# Soal 1 
# 1+3+5+...+(2n-1) = n^2

def sisi_kiri(n):
    jumlah = 0
    for i in range(1, 2*n, 2):   
        jumlah += i
    return jumlah

def sisi_kanan(n):
    return n**2
```

```python
for n in range(0, 20):
    print(n, ";", sisi_kiri(n), sisi_kanan(n), sisi_kiri(n) == sisi_kanan(n))
```

Pada output terdapat hasil perhoitungan dari sisi kiri, kemudian terdapat hasil dari perhitungan sisi kanan, dan terdapat hasil benar atau tidak dari perbandingan hasil sisi kanan dan sisi kiri. Pada output tertera jelas bahwa semua bilangan dengan n lebih dari sama dengan 0 menghasilkan angka yang valid semua.

```python
# Soal 5
# 2^n > n^2

def sisi_kiri(n):
    return 2**n

def sisi_kanan(n):
    return n**2
```
