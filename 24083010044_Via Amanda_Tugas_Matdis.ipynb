```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("data mahasiswa matdis.xlsx")
df
```

Data diatas terdiri dari 39 Mahasiswa mata kuliah matematika diskrit A dengan range data 0-1. Data diatas merupakan data menurut saya terhadap beberapa penilaian yakni adil dan amanah, suka menolong, nyambung kalau diajak ngobrol dan menarik serta asik. 

```python
print(df.columns)
```

```python
cols = ["ADIL DAN AMANAH", "SUKA MEMBANTU", 
        "NYAMBUNG KALAU NGOROL", "ASYIK DAN MENARIK"]

for c in cols:
    if df[c].isnull().all():
        df[c] = np.round(np.random.rand(len(df)), 1)

C = df[cols].to_numpy(dtype=float)
print(C)
```

Kode diatas digunakan untuk menyimpan nama-nama kolom yang akan dianalisis yang terdiri dari range 0-1 dengan kolom-kolom adil dan amanah, suka membantu, nyambung jika diajak ngobrol dan asik serta menarik. Keempat kolom ini akan dicek terlebih dahulu untuk missing values, kemudian mengubah dataframe menjadi array numpy dengan tipe data float dan akhirnya disimpan ke variabel C dan digunakan untuk perhitungan matematika. 

```python
P = np.array([0.3, 0.25, 0.25, 0.2])
P = P / P.sum()

S = C @ P
df["Skor"] = S

print(df["Skor"])
```

Kode diatas digunakan untuk membuat vektor bobot berupa array yang berisi 4 angka yang tiap elemennya menjelaskan bobot untuk kolom atau fitur (A, B, C, D) dan di normalisasi vektor bobot agar jumlahnya samadengan 1. Setelah itu, P.sum() digunakan untuk menghitung total seluruh elemen dari P yang dibagi dengan jumalh p. Alasannya adalah setelah di normalisasi p sebagai proporsi tiap fitur sehingga nanti hasil skor menjadi weighted average yang mudah untuk diinterpretasikan. 

Bila c berada di [0,1] dan p ditambahkan jadi 1, maka skor juga akan berada pada kisaran [0,1] yang artinya sebagai proporsi kedekatan atau kecocokan. Semakin mendekati 1 maka semakin cocok, begitupun sebaliknya jika mendekati 0 maka kurang cocok.

```python
top4 = df.nlargest(4, "Skor")
print("Empat mahasiswa terbaik:")
print(top4[["NAMA", "Skor"]])
```

Kode diatas digunakan untuk menampilkan 4 mahasiswa terbaik yang memiliki nilai paling tinggi. Pada hasilnya, terdapat 4 nama tertinggi yaitu Maulida, Alfani, Madina, dan Fanny dengan perbedaan rentang skor yang sedikit.

```python
cov_mat = pd.DataFrame(np.cov(C, rowvar=False, ddof=1), 
                       index=cols, columns=cols)
print("\nKovarian:\n", cov_mat)
```

Kode diatas digunakan untuk menghitung matriks kovarian dengan menganggap setiap kolom sebagai variabel dan setiap baris adalah mahasiswa dan derajat bebasnya adalah 1 yang artinya menghitung sample covariance. Pada output terlihat range [0,1] yang memiliki arti jika nilai mendekati satu maka hubungan antara kriteria satu dengan kriteria yang lain adalah kriteria kuat dan jika salah satu kriteria naik dan yang lain cenderung ikut naik atau memiliki hubungan searah. Dan jika nilai negatif, jika salah satu naik maka yang lain cenderung turun atau memiliki hubungan berlawanan. Misal pada output, adil dan amanah dengan nyambung kalau ngobrol adalah nilainya adalah 0,0726665, angka tersebut adalah nilai kovarians antara dua kriteria. Dan antara kedua kriteria tersebut memiliki nilai positif dan memiliki hubungan kriteria kuat.

```python
corr_mat = pd.DataFrame(np.corrcoef(C, rowvar=False), 
                        index=cols, columns=cols)
print("\nKorelasi:\n", corr_mat)
```

Kode diatas digunakan untuk menghitung koefisien korelasi pearson antar karakteristik di matriks C. Pada kolom C dianggap sebagai variabel, sedangkan baris adalah mahasiswa. Pada output, jika nilai positif mendekati 1 artinya hubungan positif kuat, misal pada suka membantu dan nyambung jika diajak ngobrol dengan nilai 0.698174 memiliki hubungan kuat artinya mahasiswa yang suka membantu cenderung juga mudah nyambung diajak ngobrol. Jika nilai mendekati 0, maka hubungan lemah atau hampir tidak ada hubungan linear antar variabel. Jika nilai negatif, maka hubungan berlawanan artinya jika semakin tinggi suatu variabel maka variabel lain semakin rendah. 

```python
norms = np.linalg.norm(C, axis=1, keepdims=True)
norms[norms == 0] = 1
C_norm = C / norms

cosine_sim = C_norm @ C_norm.T
cosine_df = pd.DataFrame(cosine_sim, index=df["NAMA"], columns=df["NAMA"])

print("\nCosine Similarity (potongan):\n", cosine_df.head())
```

Kode diatas digunakan untuk menghitung panjang ventor setiap mahasiswa pada matriks C dan menormalisasi setiap baris mahasiswa sehingga memiliki panjang vektor = 1. Dengan begitu, perbandingan antara mahasiswa satu dengan mahasiswa yang lain tidak hanya diperoleh dari besarnya nilai melainkan dapat dari arah vektor saja. Dari output cosine similarity yang mendekati 1 menunjukkan bahwa pola penilaian antara kedua mahasiswa hampir sama artinya karakter mereka mirip. Contoh pada madina dan aquina dengan nilai 0,99 yang artinya kedua mahasiswa tersebut memiliki sifat hampir sama dan identik dengan keempat aspek tersebut. Jika pada kedua mahasiswa memiliki nilai yang lebih kecil artinya kedua mahasiswa tersebut tidak memiliki kemiripan satu sama lain dengan emoat aspek diatas.

```python
pairs = []
for i in range(len(cosine_df)):
    for j in range(i+1, len(cosine_df)):
        pairs.append((cosine_df.index[i],
                      cosine_df.index[j],
                      cosine_df.iloc[i, j]))

# Urutkan berdasarkan similarity tertinggi
pairs_sorted = sorted(pairs, key=lambda x: x[2], reverse=True)

# Tampilkan 5 pasangan paling mirip
print("5 Pasangan Mahasiswa Paling Mirip (berdasarkan cosine similarity):")
for p in pairs_sorted[:5]:
    print(f"{p[0]} ↔ {p[1]} : {p[2]:.3f}")
```

Tujuan kode ini adalah mencari tau 5 pasangan mahasiswa yang memiliki kemiripan paling tinggi. Pada hasil output terdapat beberapa nama dengan nilai kemiripan 1. 

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(18,14))
sns.heatmap(cosine_df, annot=True, cmap="coolwarm", fmt=".2f", cbar=True)

plt.title("Cosine Similarity antar Siswa", fontsize=14, pad=15)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

Kode diatas digunakan untuk membuat heatmap dari dataframe mahasiswa dan nilai empat kategori yang merupakan nilai consine similarity. Pada visualisasi heatmap diatas, terdapat beberapa kotak yang memiliki warna merah pekat dengan nilai mendekati 1 yang artinya kemiripan sangat tinggi antara kedua mahasiswa, untuk kotak yang memiliki warna merah muda memiliki rentang 0,6-0,8 artinya terdapat sedikit kemiripan antara kedua mahasiswa. Untuk kotak warna biru atau lebih terang memiliki nilai 0,5 kebawah yang menunjukkan kemiripan lebih rendah dibandingkan dengan kotak warna merah. 

```python
from sklearn.decomposition import PCA

X = df[["ADIL DAN AMANAH", "SUKA MEMBANTU", "NYAMBUNG KALAU NGOROL", "ASYIK DAN MENARIK"]]

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

df_pca = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
df_pca["NAMA"] = df["NAMA"]

print("Hasil PCA:\n", df_pca)

plt.figure(figsize=(15,8))
plt.scatter(df_pca["PC1"], df_pca["PC2"], c="blue", s=80)

for i, nama in enumerate(df_pca["NAMA"]):
    plt.text(df_pca["PC1"][i]+0.02, df_pca["PC2"][i]+0.02, nama)

plt.title("Visualisasi PCA (2 Komponen Utama)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)
plt.show()
```

PCA atau Principal Component Analysis adalah salah satu metode reduksi dimensi dalam analisis data. PCA dipakai untuk menyedehanakan data yang memiliki banyak variabel menjadi lebih sedikit tetapi tetap mempertahankan informasi terpenting dari data tersebut. PCA akan menggabungkan hubungan antara indikator-indikator menjadi komponen utama yang disebut dengan PC atua principal component. Pada umumnya, PC1 menyimpan variasi data terbesar kemudian disusul oleh PC2. 

Pada output, jika bernilai positif pada PC1 maka cenderung berada kearah kanan dan untuk PC2 cenderung kearah atas artinya data lebih mirip dengan variabel yang memberi konstribusi positif. Dan untuk negatif pada PC1 cenderung kearah kiri dan untuk PC2 cenderung kearah bawah yang artinya data lebih mirip dengan variabel yang memberi kontribusi negatif pada komponen. Untuk titik yang berada dekat dengan 0 pada PC1 atau PC2 artinya data tidak condong kearah positif atau negatif. 

Siswa yang posisinya berdekatan adalah Febriani Yolanda Tessa dan Alimun Hakim memiliki karakteristik penilaian yang miirp. Visualisasi ini memudahkan untuk melihat pola kelompok dan outlier pada mahasiswa berdasarkan penilaian karakter. 

```python
df["RATA_RATA"] = df[cols].mean(axis=1)
df["TARGET"] = (df["RATA_RATA"] >= 0.6).astype(int)
print(df["TARGET"])
```

Kode ini digunakan untuk mengklasifikasikan beberapa data mahasiswa matdis kelas A ke dalam dua kategori yaitu 0 dan 1, berdasarkan nilai rata-ratanya. Untuk nilai 1 artinya mahasiswa tersebut dekat dengan saya sedangkan untuk 0 mahasiswa tersebut tidak dekat dengan saya.

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

df["RATA_RATA"] = df[cols].mean(axis=1)
df["TARGET"] = (df["RATA_RATA"] >= 0.6).astype(int)

X = df[cols]
y = df["TARGET"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

y_pred = model.predict(X)

print("Akurasi:", accuracy_score(y, y_pred))
print("\nLaporan klasifikasi:\n", classification_report(y, y_pred))
print("\nTarget Asli :", list(y.values))
print("Prediksi    :", list(y_pred))
```

Tujuan dari kode diatas adalah untuk membangun model klasifikasi dengan Decision Tree berdasarkan nilai rata-rata indikator mahasiswa, mengelomppokkan menjadi dua kategori dan mengevaluasi seberapa baik model dalam menirukan atau mempelajari pola dari kategori yang diberikan. 

Hasil menunjukkan bahwa model Decision Tree berhasil melakukan klasifikasi dengan hasil sempurna, ditandai oleh akurasi sebesar 100%. Pada kedua kelas, yaitu kelas 0 (tidak dekat) dan kelas 1 (dekat), nilai precision, recall, serta f1-score semuanya mencapai 1.00. Hal ini berarti tidak ada kesalahan dalam memprediksi data. Hasil prediksi dengan target sebenarnya sama, sehingga semua mahasiswa terklasifikasi sesuai dengan label yang sebenarnya. Namun, hasil yang sempurna ini juga bisa menjadi tanda adanya overfitting, terutama jika model hanya diuji pada data latih dan belum diuji dengan data uji yang benar-benar terpisah.

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

print("Hasil clusterisasi:\n", df[["NAMA", "Cluster"]])

plt.figure(figsize=(15,10))
plt.scatter(df["NYAMBUNG KALAU NGOROL"], df["SUKA MEMBANTU"], 
            c=df["Cluster"], cmap="coolwarm", s=120, edgecolors="k")

for i, nama in enumerate(df["NAMA"]):
    plt.text(df["NYAMBUNG KALAU NGOROL"][i]+0.02, df["SUKA MEMBANTU"][i]+0.02, nama, fontsize=10)

centers = kmeans.cluster_centers_
plt.scatter(centers[:,0], centers[:,1], c="yellow", s=200, marker="X", edgecolors="k", label="Centroid")

plt.title("Clusterisasi Mahasiswa", fontsize=14)
plt.xlabel("NYAMBUNG KALAU NGOROL")
plt.ylabel("SUKA MEMBANTU")
plt.legend()
plt.grid(True)
plt.show()
```

Kode ini bertujuan mengelompokkan mahasiswa menggunakan metode K-Means dengan dua kelompok. Data yang digunakan berasal dari dua aspek, yaitu "nyambung kalau ngobrol" dan "suka membantu". Setiap mahasiswa diberi label kelompok (0 atau 1) berdasarkan seberapa dekat nilai mereka dengan pusat kelompok.

Hasil menunjukkan mahasiswa terbagi menjadi dua kelompok utama berdasarkan kesamaan dalam kemampuan berbicara saat ngobrol dan sifat suka membantu. Mahasiswa yang memiliki karakteristik serupa akan masuk ke kelompok yang sama. Titik pusat (lingkaran kuning) mewakili rata-rata posisi setiap kelompok. Semakin dekat mahasiswa dengan titik pusat, semakin kuat hubungan mereka dengan karakteristik kelompok tersebut. Visualisasi ini menunjukkan hasil pengelompokan mahasiswa berdasarkan dua hal utama: "Nyambung Kalau Ngobrol" (sumbu X) dan "Suka Membantu" (sumbu Y). Titik biru dan merah masing-masing mewakili dua kelompok yang berbeda, sedangkan tanda X kuning menunjukkan pusat dari masing-masing kelompok. Mahasiswa di kelompok biru cenderung memiliki skor yang lebih tinggi pada kedua aspek, sehingga lebih ramah dalam berbicara dan lebih suportif. Sementara itu, mahasiswa di kelompok merah memiliki skor lebih rendah, sehingga tampak lebih individualis atau kurang aktif dalam membantu orang lain atau berinteraksi dengan teman.
