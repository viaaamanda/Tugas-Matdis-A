```python
import pandas as pd
import string

text_abstrak = """Permasalahan energi masih menjadi topik besar dan tantangan berat di Indonesia sehingga penting menghadirkan langkah strategis untuk mencari solusi, baik melalui penelitian maupun kajian. Lembaran Publikasi Minyak dan GasBumi, Scientific Contribution Oil and Gas, dan Jurnal Teknologi Minyak dan Gas Bumi merupakan sarana yang digunakan peneliti untuk menghimpun hasil penelitian dan kajian inovatifnya. Untuk mengetahui muatan artikel,pola kepengarangan, karakteristik, dan produktivitas peneliti digunakan kajian bibliometrik. Selama kurun waktu sebanyak peneliti menghasilkan artikel. Hasil penelitian terbanyak berada pada teknologi eksploitasi sebesar artikel, sedangkan fokus utama lain, seperti teknologi eksplorasi mencapai artikel.Jangkauan kerja sama penelitian mencakup antarnegara, melibatkan dua hingga empat peneliti lebih dengan tingkat kolaborasi mencapai. Keseluruhan aktivitas penelitian minyak dan gas migas didominasi institusi penelitiandan pengembangan litbang, perguruan tinggi, dan pelaku bisnis."""

vowels = "aiueoAIUEO"

translator = str.maketrans('', '', string.punctuation)
clean_text_abstrak = text_abstrak.translate(translator)

words = clean_text_abstrak.split()

data = []
for word in words:
    num_vowels = sum(1 for ch in word if ch in vowels)
    num_consonants = sum(1 for ch in word if ch.isalpha() and ch not in vowels)
    data.append({"Kata": word, "Vokal": num_vowels, "Konsonan": num_consonants})

df1 = pd.DataFrame(data)
df1
```

Penjelasan : Kode pivot freq digunakan untuk menghitung berapa banyak kata yang memiliki kombinasi tertentu antara jumlah vokal dan konsonan pada df_all. ```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,6))
sns.heatmap(pivot_freq, annot=True,fmt="d", cmap="YlOrRd",  linewidths=0.5, linecolor='white', cbar_kws={'label': 'Frekuensi'})

plt.title("Heatmap Frekuensi Kombinasi Vokal dan Konsonan pada Kata", fontsize=13, fontweight='bold')
plt.xlabel("Jumlah Konsonan", fontsize=11)
plt.ylabel("Jumlah Vokal", fontsize=11)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.show()
```

Penjelasan : Jadi kode diatas digunakan untuk membuat pivot tabel yang menunjukkan jumlah kata unik atau tidak duplikat berdasarkan kombinasi jumlah vokal dam konsonan. ```python
cacah_kombinasi = df_all.groupby(['Vokal', 'Konsonan']).size().reset_index(name='Cacah_Kata')
print(cacah_kombinasi)
```

Penjelasan : Kode diatas digunakan untuk melihat pola distribusi kata berdasarkan jumlah vokal dan konsonan dengan menentukan kombinasi antara huruf vokal dan konsonan. Pada output kolom cacah kata berisi jumlah kata yang termasuk dalam kombinasi tertentu dari banyaknya vokal dan konsonan. Output juga menunjukkan ada 13 kata yang memiliki 1 vokal dan 1 konsonan, terdapat 39 kata yang memiliki 1 vokal dan 2 konsonan, dll. 

```python
import plotly.express as px

fig = px.scatter_3d(
    cacah_kombinasi,
    x='Vokal',
    y='Konsonan',
    z='Cacah_Kata',
    color='Cacah_Kata',
    size='Cacah_Kata',
    color_continuous_scale='Viridis',
    title='Grafik 3D Jumlah Kata berdasarkan Vokal dan Konsonan'
)
fig.update_traces(marker=dict(opacity=0.8))
fig.show()
```

Penjelasan : Kode diatas digunakan untuk memvisualisasikan grafik 3D yang menunjukkan distribusi jumlah kata berdasarkan informasi pada output kombinasi jumlah vokal dan jumlah konsonan dengan sumbu x sebagai vokal, sumbu y sebagai konsonan dan sumbu z sebagai cacah kata. Dari visualisasi terlihat bahwa titik dengan frekuensi kata terbesar yang berwarna kuning dan memiliki ukuran bubble paling besar berada pada kombinasi jumlah vokal 2-3 dan konsonan 3-4 yang artinya mayoritas kata dalam dataset cenderung memiliki 2 hingga 3 vokal dan 3 hingga 4 konsonan. Untuk bubble kecil berwarna ungu menunjukkan kombinasi vokal dan konsonan yang terlalu kecil dengan jumlah kata yang ditemukam relatif sedikit. Jadi hal ini menunjukkan bahwa terdapat struktur kata yang paling umum dalam dataset berada pada panjang kata. 

```python
cacah_kombinasi_sorted = cacah_kombinasi.sort_values(by=['Vokal', 'Konsonan'])

plt.figure(figsize=(10,6))
sns.lineplot(
    data=cacah_kombinasi_sorted, 
    x=range(len(cacah_kombinasi_sorted)), 
    y='Cacah_Kata', 
    marker='o'
)

plt.title("Line Plot Cacah Kata Berdasarkan Kombinasi Vokal-Konsonan (Ordered Data)")
plt.xlabel("Urutan Kombinasi (berdasarkan Vokal & Konsonan)")
plt.ylabel("Jumlah Cacah Kata")
plt.grid(True)
plt.show()
```

Pertama, kode tersebut mengimpor daftar kata-kata tidak penting dalam bahasa Indonesia dari nltk, seperti "dan", "di", atau "yang", yang biasanya tidak memiliki makna spesifik dalam analisis.
